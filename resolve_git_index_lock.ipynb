```python
This notebook guides you through resolving a stale Git lock file that blocks commits in a repository.

---

## 1. Check for Running Git Processes
Use Task Manager or command-line tools to find and stop any active Git processes that might be holding the repository lock.

**Example:**
- Windows: open Task Manager and end `git.exe`, `gitk.exe`, or editors that may have started Git.
- Command line: `tasklist /FI "IMAGENAME eq git.exe"`.

---

## 2. Inspect the Repository for Lock Files
Check the `.git` directory for the existence of `index.lock` or other temporary lock files.

**Example:**
- `dir .git\index.lock`
- `dir .git\*.lock`

---

## 3. Remove a Stale `.git/index.lock` File
If no Git processes are running, delete the stale lock file manually.

**Example:**
- `del .git\index.lock`

---

## 4. Verify Repository Status
Run `git status` to confirm the repository is clean and no lock issues remain.

**Example:**
- `git status`

---

## 5. Retry the Git Commit
Attempt the commit again after resolving the lock file issue.

**Example:**
- `git commit -m "Set Node engine to 24.x for Vercel"`
```
